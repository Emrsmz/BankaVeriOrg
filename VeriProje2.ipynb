```python
import numpy as np
import pandas as pd
durum = 1
number = 0
erken = True
```

```python
def dosyaoku(dosya):
    if "0" == dosya:
        print("Islem bitirilmiştir.")
        erken = False
    elif ".xlsx" in dosya:
        veri0 = pd.read_excel(dosya)
        return veri0
    elif ".csv" in dosya:
        veri0 = pd.read_csv(dosya)
        return veri0
veri1 = dosyaoku("Readingdata/hesaphareketleri.xlsx")
```

```python
def basitduzenle(veri1):
    veri1 = veri1.drop(veri1.index[0:14])
    veri2 = veri1.rename(columns={"Unnamed: 0":"Tarih","Unnamed: 2":"Etiket","Unnamed: 3":"Tutar","Unnamed: 4":"Bakiye","Unnamed: 5":"Dekont No","T. GARANTİ BANKASI A.Ş.\nGenel Müdürlük: Nispetiye Mah. Aytar Cad.No: 2, Beşiktaş, 34340, Levent, İstanbul\nBüyük Mükellefler Vergi Dairesi Başkanlığı Vergi No: 8790017566\nMersis Numarası: 0879 0017 5660 0379\nwww.garantibbva.com.tr":"Açıklama"})
    veri2 = veri2.reset_index(drop=True)
    veri2 = veri2.drop(columns = ["Bakiye","Dekont No"])
    return veri2
```

```python
def topgelir(df):
    gel1 = int(df[df["Tutar"] > 0]["Tutar"].sum())
    return gel1
def topgider(df):
    git1 = int(df[df["Tutar"] < 0]["Tutar"].sum())
    return git1
```

```python
kurallar = {
    "kahve": ["starbucks", "cafe", "kahve","gloria","burfas","cay","kafe","coffee","umay"],
    "yemek": ["mc","burger","pizza","doner","döner","pilav","durum","kitchen","tavuk","tavu","popeyes"],
    "alışveriş": ["trendyol", "amazon", "alışveriş","mrdiy","koctas"],
    "ulaşım": ["uber", "obilet", "taxi","enuygun"],
    "egitim": ["kitap","udemy"],
    "mobil": ["mobil","vodafone","turkcell","telekom"],
    "aidat": ["aidat"],
    "spotify": ["spotify"],
    "market" : ["migros","bim","market","carrefour","happy","gida","gıda","a101"]
}

def kategori(text):
    text = text.lower().replace("i̇", "i")

    for kat, keywords in kurallar.items():
        if any(k in text for k in keywords):
            return kat

    return "diğer"
```

```python
def bosluk():
    print("\n================\n")
```

```python
while durum == 1:

    print("Taranmasını istediğiniz dökümanın yolunu giriniz:")
    print("(.xlsx veya .csv türünde olmalıdır.)")

    dosya = input("Dosya yolu: ")
    veri1 = dosyaoku(dosya)
    
    while True:
        print("Yapmak istediğiniz işlemi seçiniz:")
        print("1 - Veri göster.")
        print("2 - Veri düzenle")
        print("3 - Veri toplam gelir - gider")
        print("0 - Işlemi bitirin...")
        bosluk()
    
        islem = int(input("Seçim: "))
        if islem == 1:
            number = input("Gösterilecek Satır Sayısı :")
            number = int(number)
            print("\n===== VERİ =====")
            if number != 0:
                print(veri1.head(number).to_string(index=False))
            else:
                print(veri1.to_string(index=False))
                
            bosluk()
        elif islem == 2:
            veri1 = basitduzenle(veri1)
            veri1["Kategori"] = veri1["Açıklama"].apply(kategori)
        elif islem == 3:
            print("Toplam gelir :",topgelir(veri1))
            print("Toplam gider :",topgider(veri1))
            print("Toplam net : ",(topgelir(veri1)+topgider(veri1)))
            bosluk()
            
        elif islem == 0:
            durum = 0
            break
    
        else:
            print("Hata oluştu.")
            

if durum == 0:
    print("İşleminiz bitirilmiştir :)")
```
